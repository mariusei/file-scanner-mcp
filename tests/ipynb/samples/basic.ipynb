# Sales ledger

Reads the nightly export and summarises it per region.

```python
import statistics


def total(amounts):
    """Sum of the amounts."""
    return sum(amounts)


class Ledger:
    """Rows of (region, amount)."""

    def __init__(self, rows):
        self.rows = rows

    def mean(self):
        return statistics.mean(amount for _, amount in self.rows)
```
